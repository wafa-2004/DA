```python
#write a pgm to check if a set is a subset of another set
a={1,2}
b={1,2,3,4}
if a.issubset(b):
    print("a is the subset of b.")
else:
    print("a is not the subset of b.")
```

```python
#write a pgm to find the index of an element in a tuple
t=(10,20,30,40)
i=20
if i in t:
    index=t.index(i)
    print(f"the index of the {i} is {index}.")
else:
    print(f"the{i} is not present in the tuple.")
```

```python
#write a pgm to sort a list into ascending order
numbers=[12,4,56,7,23]
numbers.sort()
print("The sorted list is:",numbers)
```

```python
#write a pgm to reverse a string
text="hello"
reverse=""
i=len(text)-1
while i>=0:
    if i>=0:
        reverse+=text[i]
        i-=1
print("the reversed string is:",reverse)
```

```python
#write a pgm to get the keys and values of a dictionary separately
#name=john,age=21,grade=A
std={"Name":"John","Age":21,"Grade":"A"}
keys=std.keys()
print("keys:",tuple(keys))
values=std.values()
print("value:",tuple(values))
```

```python
#write a program to remove duplicates from a list
numbers=[1,2,2,3,4,4,5]
my_list=[]
for i in numbers:
    if i not in my_list:
        my_list.append(i)
print("the list after removing duplicates:",my_list)
```
