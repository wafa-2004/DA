```python
#parent clas person-name , age
#child class student-student_id
#write a mehod to display all details

class Parent:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def info(self):
        return f"I'm {self.name}. {self.age} years old."


class Student(Parent):
    def __init__(self, name, age, student_id):
        super().__init__(name, age)
        self.student_id = student_id

    def info(self):
        return f"I am {self.name}. I'm {self.age} years old. My ID number is {self.student_id}."


# Create objects
my_parent = Parent("Sakkir", 50)
my_student = Student("Wafa", 21, 61)

# Print information
print(my_parent.info())
print(my_student.info())
```
